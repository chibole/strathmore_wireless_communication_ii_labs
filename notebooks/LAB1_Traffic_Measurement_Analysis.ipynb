### 1. Introduction

Telecommunications traffic measurement is an important activity in the planning, operation and optimisation of cellular communication networks.

In a mobile cellular network, subscribers generate traffic whenever they initiate or receive calls. The network must provide sufficient communication resources to accommodate this demand while maintaining an acceptable quality of service.

Traffic engineering involves measuring call activity, determining how long communication resources remain occupied, and estimating the traffic intensity offered to the network.

In this laboratory exercise, Python will be used to organise, analyse and visualise telecommunications traffic measurements. The analysis will focus on call arrivals, call holding times, traffic volume and traffic intensity expressed in Erlangs.

The concepts introduced in this laboratory will provide the foundation for subsequent work on blocking probability, Erlang B analysis and GSM network capacity planning.

### 2. Purpose of the Laboratory

To introduce telecommunications traffic measurement and analysis, demonstrating how call activity, holding time and traffic intensity are used in planning and dimensioning cellular voice networks.

### 3. Learning Outcomes

By the end of this laboratory exercise, students should be able to:

1. Explain traffic volume, call arrival rate, call holding time and traffic intensity.
2. Calculate offered traffic in Erlangs from call and holding-time measurements.
3. Organise and analyse a telecommunications traffic dataset using Python.
4. Plot traffic variation over time and identify periods of higher and lower demand.
5. Explain how measured traffic informs GSM network capacity planning.
