### 4.1 Call Arrival Rate

The call arrival rate represents the number of calls initiated or received within a specified observation period.

It is commonly represented by the Greek letter lambda (λ).

The call arrival rate is calculated as:

**λ = N/T**

Where:

- λ = Call arrival rate
- N = Total number of calls observed
- T = Observation period

The arrival rate may be expressed in calls per second, calls per minute or calls per hour.

**Worked Example 1: Calculating Call Arrival Rate**

A cellular network records 240 calls during a one-hour observation period.

Calculate the average call arrival rate.

**Solution**

Given:

- Number of calls, N = 240
- Observation period, T = 1 hour

Therefore:

λ = 240/1

**λ = 240 calls/hour**

This means that, on average, 240 calls arrive at the network every hour during the observation period.

### 4.2 Call Holding Time

Call holding time is the duration for which a communication channel or network resource remains occupied by a call.

It is commonly represented by the letter h.

For an individual call:

**h = End time − Start time**

The mean call holding time is calculated as:

**Mean holding time = Total call duration / Number of calls**

If the total duration is measured in seconds, the mean holding time will also be expressed in seconds.

**Worked Example 2: Calculating Mean Holding Time**

A cellular network records five calls with the following durations:

| Call | Duration |
|---|---:|
| 1 | 60 seconds |
| 2 | 120 seconds |
| 3 | 180 seconds |
| 4 | 90 seconds |
| 5 | 150 seconds |

Calculate the mean call holding time.

**Solution**

Total call duration:

60 + 120 + 180 + 90 + 150 = 600 seconds

Number of calls = 5

Mean holding time:

h = 600/5

**h = 120 seconds**

Therefore, the average call occupies a communication resource for 120 seconds, equivalent to 2 minutes.

### 4.3 Traffic Volume

Traffic volume represents the total accumulated duration of communication activity during an observation period.

It is calculated by adding the holding times of all calls observed during that period.

**V = h₁ + h₂ + h₃ + ... + hₙ**

Where:

- V = Total traffic volume
- h₁, h₂, ... hₙ = Individual call holding times
- n = Total number of calls

Traffic volume is commonly expressed in call-seconds, call-minutes or call-hours.

**Worked Example 3: Calculating Traffic Volume**

A cellular network records 120 calls during an observation period. The average call duration is 2 minutes.

Calculate the total traffic volume.

**Solution**

Total traffic volume:

V = Number of calls × Mean holding time

V = 120 × 2

**V = 240 call-minutes**

Converting into hours:

V = 240/60

**V = 4 call-hours**

The network has therefore accumulated 240 minutes of call activity during the observation period.

### 4.4 Traffic Intensity

Traffic intensity represents the average number of communication resources occupied during a specified observation period.

It is measured in Erlangs and is commonly represented by the letter A.

Traffic intensity is calculated using:

**A = V/T**

Alternatively, when the call arrival rate and mean holding time are known:

**A = λh**

Where:

- A = Traffic intensity in Erlangs
- V = Total traffic volume
- T = Observation period
- λ = Call arrival rate
- h = Mean call holding time

For the equation A = λh, the time units used for λ and h must be consistent.

One Erlang represents an average traffic load equivalent to one communication resource being continuously occupied throughout the observation period.

For example, 4 Erlangs represents an average offered traffic load equivalent to four continuously occupied channels.

**Worked Example 4: Calculating Traffic Intensity**

A cellular network receives 120 calls per hour. The average call holding time is 2 minutes.

Calculate the offered traffic intensity.

**Solution**

Given:

- Call arrival rate = 120 calls/hour
- Mean holding time = 2 minutes

First, convert the mean holding time into hours:

h = 2/60

h = 0.03333 hours

Using:

A = λh

A = 120 × 0.03333

**A = 4 Erlangs**

Therefore, the offered traffic intensity is approximately 4 Erlangs.

This means that the network experiences an average traffic load equivalent to four continuously occupied communication channels.

### 4.5 Relationship Between Traffic Volume and Traffic Intensity

Traffic volume and traffic intensity are related, but they describe different aspects of telecommunications traffic.

Traffic volume measures the total accumulated call duration, whereas traffic intensity measures the average traffic load during the observation period.

The relationship is:

**Traffic Intensity = Traffic Volume / Observation Time**

For example, if a network records 240 call-minutes during a 60-minute observation period:

A = 240/60

**A = 4 Erlangs**

If the same 240 call-minutes are accumulated over 120 minutes:

A = 240/120

**A = 2 Erlangs**

Although the total traffic volume is identical, the average traffic intensity is different because the observation periods are different.

### 4.6 Important Considerations in Traffic Measurement

When analysing telecommunications traffic, the following considerations are important:

1. **Observation period:** Traffic measurements must be associated with a clearly defined observation period.
2. **Time consistency:** Call arrival rate and holding time must use compatible time units when calculating traffic intensity.
3. **Busy hour:** Traffic demand varies throughout the day. The busy hour is the one-hour period with the highest measured traffic intensity.
4. **Traffic fluctuations:** Average traffic intensity does not describe every instantaneous change in network demand.
5. **Network capacity:** The number of channels required depends not only on offered traffic but also on the acceptable call-blocking probability and the traffic model.

These concepts will be applied in the subsequent Python-based analysis.

### 5.4 Generating Your Personal Dataset

The simulation will generate call arrivals throughout a 24-hour period. Call arrivals will vary according to the specified daily traffic profile, representing lower overnight activity and higher demand during the daytime and evening.

Call durations will be generated using a random statistical distribution.

Create another **Code cell**, copy the complete code below and execute it.

```python
# ============================================================
# BEE 5224 - WIRELESS COMMUNICATIONS II
# LAB 1: TELECOMMUNICATIONS TRAFFIC MEASUREMENT AND ANALYSIS
# PERSONAL TRAFFIC DATASET GENERATION
# ============================================================

# 1. Generate a reproducible random seed from your admission number

digest = hashlib.sha256(
    admission_number.encode("utf-8")
).digest()

random_seed = int.from_bytes(digest[:4], byteorder="big")

rng = np.random.default_rng(random_seed)

print("Your simulation seed:", random_seed)

# 2. Define the observation period

observation_date = pd.Timestamp("2026-03-17")

# Expected call arrival profile for the 24 hours.
# Higher values represent periods of greater expected demand.

hourly_call_profile = np.array([
    25, 15, 10, 8, 7, 10,
    20, 35, 50, 60, 65, 65,
    60, 55, 55, 70, 100, 160,
    110, 80, 60, 45, 35, 25
], dtype=float)

# Scale the profile to approximately 1,250 calls per day.

target_daily_calls = 1250

hourly_means = (
    hourly_call_profile
    * target_daily_calls
    / hourly_call_profile.sum()
)

# 3. Generate individual call records

records = []
call_counter = 1

for hour in range(24):

    # Generate the number of calls arriving during this hour.
    number_of_calls = rng.poisson(hourly_means[hour])

    for _ in range(number_of_calls):

        # Select a random arrival time within the hour.
        seconds_after_hour = rng.integers(0, 3600)

        start_time = (
            observation_date
            + pd.Timedelta(hours=hour)
            + pd.Timedelta(seconds=int(seconds_after_hour))
        )

        # Generate call duration in seconds.
        # The underlying exponential distribution has a
        # scale parameter of 120 seconds.
        duration_sec = int(
            np.clip(rng.exponential(scale=120), 15, 1800)
        )

        end_time = start_time + pd.Timedelta(
            seconds=duration_sec
        )

        records.append({
            "call_id": f"CALL_{call_counter:05d}",
            "start_time": start_time,
            "end_time": end_time,
            "duration_sec": duration_sec,
            "hour": hour
        })

        call_counter += 1

# 4. Organise the generated records into a DataFrame

traffic_data = pd.DataFrame(records)

traffic_data = traffic_data.sort_values(
    "start_time"
).reset_index(drop=True)

# 5. Create a data folder and save the CSV file

data_folder = Path("data")
data_folder.mkdir(exist_ok=True)

output_file = data_folder / "traffic_measurements.csv"

traffic_data.to_csv(output_file, index=False)

# 6. Display the generation summary

print("\nDATASET GENERATION COMPLETE")
print("------------------------------------------")
print("Admission number:", admission_number)
print("Random seed:", random_seed)
print("Observation date:", observation_date.date())
print("Total number of calls:", len(traffic_data))
print(
    "Mean call duration:",
    round(traffic_data["duration_sec"].mean(), 2),
    "seconds"
)
print(
    "Total call duration:",
    traffic_data["duration_sec"].sum(),
    "seconds"
)
print("Saved file:", output_file)

print("\nFirst five call records:")
display(traffic_data.head())
```
